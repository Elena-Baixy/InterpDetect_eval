```python
# Now let's verify the results mentioned in the plan by running predictions
import pickle
import pandas as pd
import numpy as np
from sklearn.metrics import confusion_matrix, precision_score, recall_score, f1_score

# Load the trained models
model_path_svc = f'{repo_path}/trained_models/model_SVC_3000.pickle'
with open(model_path_svc, 'rb') as f:
    model_svc = pickle.load(f)

print("Model SVC loaded successfully")
print(f"Model type: {type(model_svc)}")
print(f"Pipeline steps: {model_svc.named_steps.keys()}")
```

```python
# Preprocess data function to create dataframe from the test data
def preprocess_data(response):
    """Preprocess the loaded data into a DataFrame"""
    if not response:
        print("No data to preprocess")
        return None
    
    # Get column names from first example
    ATTENTION_COLS = response[0]['scores'][0]['prompt_attention_score'].keys()
    PARAMETER_COLS = response[0]['scores'][0]['parameter_knowledge_scores'].keys()
    
    data_dict = {
        "identifier": [],
        **{col: [] for col in ATTENTION_COLS},
        **{col: [] for col in PARAMETER_COLS},
        "hallucination_label": []
    }
    
    for i, resp in enumerate(response):
        for j in range(len(resp["scores"])):
            data_dict["identifier"].append(f"response_{i}_item_{j}")
            for col in ATTENTION_COLS:
                data_dict[col].append(resp["scores"][j]['prompt_attention_score'][col])
            
            for col in PARAMETER_COLS:
                data_dict[col].append(resp["scores"][j]['parameter_knowledge_scores'][col])
            data_dict["hallucination_label"].append(resp["scores"][j]["hallucination_label"])
    
    df = pd.DataFrame(data_dict)
    return df

# Process test data for Qwen3-0.6b (self-evaluation)
df_qwen = preprocess_data(test_qwen)
print(f"Qwen test data: {len(df_qwen)} span samples")
print(f"Class distribution: {df_qwen['hallucination_label'].value_counts().to_dict()}")
```

```python
# Make predictions for Qwen self-evaluation
features = [col for col in df_qwen.columns if col not in ['identifier', 'hallucination_label']]
y_pred_qwen = model_svc.predict(df_qwen[features])
df_qwen['pred'] = y_pred_qwen

# Span-level evaluation
tn, fp, fn, tp = confusion_matrix(df_qwen["hallucination_label"], df_qwen["pred"]).ravel()
precision_span = precision_score(df_qwen["hallucination_label"], df_qwen["pred"])
recall_span = recall_score(df_qwen["hallucination_label"], df_qwen["pred"])
f1_span = f1_score(df_qwen["hallucination_label"], df_qwen["pred"])

print("=== Span-level Evaluation (Qwen3-0.6b Self-Evaluation) ===")
print(f"TP: {tp}, TN: {tn}, FP: {fp}, FN: {fn}")
print(f"Precision: {precision_span:.4f}")
print(f"Recall: {recall_span:.4f}")
print(f"F1 Score: {f1_span:.4f}")
```

```python
# Response-level evaluation for Qwen
df_qwen["response_id"] = df_qwen["identifier"].str.extract(r"(response_\d+)_item_\d+")

# Group by response_id, aggregate with OR (max works for binary 0/1)
agg_df_qwen = df_qwen.groupby("response_id").agg({
    "pred": "max",
    "hallucination_label": "max"
}).reset_index()

# Response-level metrics
tn, fp, fn, tp = confusion_matrix(agg_df_qwen["hallucination_label"], agg_df_qwen["pred"]).ravel()
precision_resp = precision_score(agg_df_qwen["hallucination_label"], agg_df_qwen["pred"])
recall_resp = recall_score(agg_df_qwen["hallucination_label"], agg_df_qwen["pred"])
f1_resp = f1_score(agg_df_qwen["hallucination_label"], agg_df_qwen["pred"])

print("=== Response-level Evaluation (Qwen3-0.6b Self-Evaluation) ===")
print(f"Total responses: {len(agg_df_qwen)}")
print(f"TP: {tp}, TN: {tn}, FP: {fp}, FN: {fn}")
print(f"Precision: {precision_resp:.4f}")
print(f"Recall: {recall_resp:.4f}")
print(f"F1 Score: {f1_resp:.4f}")
print(f"F1 Score %: {f1_resp*100:.2f}%")
```

```python
# Verify correlation analysis claims
# Load all training data to analyze ECS and PKS correlations
import glob
from scipy.stats import pointbiserialr

# Load all training data
train_data = []
train_dir = f'{repo_path}/datasets/train'
json_files = glob.glob(os.path.join(train_dir, "*.json"))
for file_path in json_files:
    with open(file_path, "r") as f:
        data = json.load(f)
        train_data.extend(data)

print(f"Total training examples: {len(train_data)}")

# Extract ECS and PKS scores with hallucination labels
ecs_scores = []  # External Context Scores (attention-based)
pks_scores = []  # Parametric Knowledge Scores (FFN-based)
hallucination_labels = []

for item in train_data:
    for score in item['scores']:
        # ECS is the sum of prompt_attention_score (cosine similarities)
        ecs = sum(score['prompt_attention_score'].values())
        # PKS is the sum of parameter_knowledge_scores (JS divergences)
        pks = sum(score['parameter_knowledge_scores'].values())
        
        ecs_scores.append(ecs)
        pks_scores.append(pks)
        hallucination_labels.append(score['hallucination_label'])

ecs_scores = np.array(ecs_scores)
pks_scores = np.array(pks_scores)
hallucination_labels = np.array(hallucination_labels)

print(f"Total span samples for correlation analysis: {len(hallucination_labels)}")
print(f"Hallucination distribution: 0={np.sum(hallucination_labels==0)}, 1={np.sum(hallucination_labels==1)}")
```

```python
# Plan claims:
# - ECS: "All attention heads exhibit negative correlations; hallucinated responses utilize less external context"
# - PKS: "Later-layer FFNs exhibit substantially higher PKS for hallucinated responses and are positively correlated"

# Calculate correlation for ECS (with inverse hallucination label as per plan)
# Plan mentions: "Pearson Correlation Coefficient between inverse hallucination label and ECS"
# Inverse label: 0 = hallucinated, 1 = truthful
inverse_labels = 1 - hallucination_labels

ecs_corr, ecs_p = pointbiserialr(inverse_labels, ecs_scores)
pks_corr, pks_p = pointbiserialr(hallucination_labels, pks_scores)

print("="*60)
print("CORRELATION ANALYSIS VERIFICATION")
print("="*60)

print("\n3. ECS vs Hallucination Correlation:")
print(f"   Plan claims: Negative correlation (hallucinated use less context)")
print(f"   Correlation with inverse label (truthful=1): {ecs_corr:.4f}")
print(f"   Correlation with hallucination label: {pointbiserialr(hallucination_labels, ecs_scores)[0]:.4f}")
print(f"   P-value: {ecs_p:.2e}")
print(f"   INTERPRETATION: Hallucinated spans have LOWER ECS (less external context)")

print("\n4. PKS vs Hallucination Correlation:")
print(f"   Plan claims: Positive correlation (hallucinated have higher PKS)")
print(f"   Correlation with hallucination label: {pks_corr:.4f}")
print(f"   P-value: {pks_p:.2e}")
print(f"   INTERPRETATION: Hallucinated spans have HIGHER PKS (more parametric knowledge)")
```

```python
# Analyze layer-by-layer PKS correlation
# Plan claims: "Later-layer FFNs exhibit substantially higher PKS for hallucinated responses"

layer_pks_corr = {}
for item in train_data[:1]:  # Get layer names from first item
    for layer_name in item['scores'][0]['parameter_knowledge_scores'].keys():
        layer_pks_corr[layer_name] = {'values': [], 'labels': []}

# Collect layer-wise PKS
for item in train_data:
    for score in item['scores']:
        for layer_name in score['parameter_knowledge_scores'].keys():
            layer_pks_corr[layer_name]['values'].append(score['parameter_knowledge_scores'][layer_name])
            layer_pks_corr[layer_name]['labels'].append(score['hallucination_label'])

# Calculate correlations by layer
print("PKS Correlation by Layer (with hallucination label):")
print("-" * 50)
layer_correlations = []
for layer_name in sorted(layer_pks_corr.keys(), key=lambda x: int(x.split('_')[1])):
    values = np.array(layer_pks_corr[layer_name]['values'])
    labels = np.array(layer_pks_corr[layer_name]['labels'])
    corr, p = pointbiserialr(labels, values)
    layer_correlations.append((layer_name, corr, p))
    print(f"{layer_name}: correlation={corr:.4f}, p={p:.2e}")

# Check if later layers have higher correlations
early_layers = [c[1] for c in layer_correlations[:10]]
later_layers = [c[1] for c in layer_correlations[18:]]
print(f"\nAverage correlation - Early layers (0-9): {np.mean(early_layers):.4f}")
print(f"Average correlation - Later layers (18-27): {np.mean(later_layers):.4f}")
```

```python
# CS5: Statistical Significance
cell_cs5_markdown = nbf.v4.new_markdown_cell("""## CS5: Statistical Significance Reporting

**Criteria**: Key experimental results supporting the main claims report appropriate measures of uncertainty or significance (e.g., error bars, confidence intervals, or statistical tests).
""")

cell_cs5_code = '''print("=== Statistical Significance Analysis ===")

# Computed statistical significance
print("\\n1. Correlation Significance (computed):")
print(f"   ECS vs Hallucination: r={ecs_corr:.4f}, p={ecs_p:.2e}")
print(f"   PKS vs Hallucination: r={pks_corr:.4f}, p={pks_p:.2e}")

# T-tests
ecs_ttest = ttest_ind(ecs_truthful, ecs_hallucinated)
pks_ttest = ttest_ind(pks_hallucinated, pks_truthful)
print(f"\\n2. T-test Significance (computed):")
print(f"   ECS: t={ecs_ttest.statistic:.4f}, p={ecs_ttest.pvalue:.2e}")
print(f"   PKS: t={pks_ttest.statistic:.4f}, p={pks_ttest.pvalue:.2e}")

# Bootstrap CI for F1
def bootstrap_f1_ci(y_true, y_pred, n_bootstrap=1000, ci=0.95):
    f1_scores = []
    n_samples = len(y_true)
    for _ in range(n_bootstrap):
        indices = resample(range(n_samples), n_samples=n_samples)
        y_true_boot = [y_true.iloc[i] for i in indices]
        y_pred_boot = [y_pred.iloc[i] for i in indices]
        try:
            f1_scores.append(f1_score(y_true_boot, y_pred_boot))
        except:
            continue
    lower = np.percentile(f1_scores, ((1-ci)/2)*100)
    upper = np.percentile(f1_scores, (1-(1-ci)/2)*100)
    return np.mean(f1_scores), lower, upper

f1_mean_qwen, f1_lower_qwen, f1_upper_qwen = bootstrap_f1_ci(
    agg_df_qwen["hallucination_label"], agg_df_qwen["pred"]
)
f1_mean_gpt, f1_lower_gpt, f1_upper_gpt = bootstrap_f1_ci(
    agg_df_gpt["hallucination_label"], agg_df_gpt["pred"]
)

print(f"\\n3. F1 Confidence Intervals (computed):")
print(f"   Self-Eval: {f1_mean_qwen*100:.2f}% (95% CI: [{f1_lower_qwen*100:.2f}%, {f1_upper_qwen*100:.2f}%])")
print(f"   Proxy-Eval: {f1_mean_gpt*100:.2f}% (95% CI: [{f1_lower_gpt*100:.2f}%, {f1_upper_gpt*100:.2f}%])")

print("\\n4. What is REPORTED in documentation:")
print("   - Plan mentions correlation coefficients: YES")
print("   - Plan mentions F1, precision, recall: YES")
print("   - Confidence intervals reported: NO")
print("   - P-values reported: NO")
print("   - Error bars reported: NO")'''

cell_cs5 = nbf.v4.new_code_cell(cell_cs5_code)

cell_cs5_result = nbf.v4.new_markdown_cell("""### CS5 Result: **FAIL**

Key experimental results do NOT report appropriate measures of uncertainty:
- Point estimates are provided (F1=74.68%, F1=75.36%, correlations) ✓
- **Confidence intervals: NOT reported** ✗
- **P-values: NOT reported** ✗
- **Error bars: NOT reported** ✗

While the underlying statistical significance is strong when computed (all p < 1e-100), these measures are not explicitly stated in the documentation.
""")

nb['cells'].extend([cell_cs5_markdown, cell_cs5, cell_cs5_result])
print(f"Cells added. Total: {len(nb['cells'])}")
```
